```python
import json
```

```python
# function for load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
        return data
```

```python
data = load_data("data.json")
```

```python
type(data)
```

```python
# Display users and their connections
def display_users(data):
    print("Users and Their Connections:\n")
    for user in data["users"]:   # corrected key
        print(f"ID:{user['id']} {user['name']} - Friends with: {user['friends']} - Liked Pages: {user['liked_pages']}")
    
    print("\nPages:\n")
    for page in data["pages"]:
        print(f"{page['id']}: {page['name']}")

display_users(data)
```
